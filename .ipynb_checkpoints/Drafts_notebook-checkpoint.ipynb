**Strings** are like arrays of letters so we can pick any desired letter. We can use a semicolon to select more than one consecutive elements: the first number represents the first number we want, the last is the number of the first element we ignore. The function *len* finds the length of the string. Note that numbers are not strings! So we can't select the second digit as if we were selecting the second letter of a text string. Apparently strings can not only be added but also multiplied by a number resulting in a repetition of the same string.

```python
text = 'Lorem ipsum dolor sed etiam'
print(text[3])
print(text[0:4]) #Element 4 would be the fith element, the letter m, but it is ignored
print(text[4:]) #Selects all the following characters starting from element 4
print(text[:7]) #All the previous characters
print(text[:]) #The whole string
print(text[2:-7]) #removes the last 7 characters of the string
print(text[0:5]*3)
print('String length:',len(text))

a=123
#print(a[1]) #will lead to an error
```

**Values types** can be integer, float or string. Only strings have a length, only strings can be concatenated. To make them talk to each other we first need to convert them: it can go both way from integer/float to str and vice versa. Raising to the power is performe by double ast **

```python
a=-12
b=1.4
c='2'
d='2.2'
print(type(a),type(b),type(c))

print(a+int(c))
print(str(a)+c)

print(b+float(d))
```

**Divisions** can be performed with /; additionally there is // which also floors the result to the preceeding integer or % which evaluates the remainder of an integer division. Notice that even if the actual result of a division is an integer, it will be treated as a float value

```python
a=52
b=51
c=3
d=2.7

print(a/c, b/c, b/d)
print(int(d)) #notice that when casting a float to an int it gets automatically rounded to the preceeding integer, not to the closest one!
```

**Complex numbers** are supported in Python and the imaginary unit is *j*. They belong to a different class called coomplex. Apparently the imaginary unit can not be written alone but always after a number (float or int). Curiously, complex numbers are always printed surrounded by round parentheses.

```python
print(4+1j)
print(type(4.2+1.2j))
print(4.2+1.2j)
#print(4+j) #Never alone
```

**Min and max** functions can order numbers and strings as well but not a mixture of the two. Other useful functions are **round** which can be given a second input specifying how many digits we want to keep after the dot. In any case, each function has its own description, accessible typing help(*function*) (Python explanation) or *function*? (Jupyter notebook explanation).

```python
print(min(1,5,-12.234))
print(round(-12.234))
print(round(-12.234,1))

text = 'abc'
print(max(text)) #so max and min see a string as a collection of characters if it has just one string
print(max(text,'nice','notsonice','easy')) #if the function is provided more strings to compare, it will just focus on the first letter of each string
```

**Libraries** have to be imported before being used

```python
import math
print(math.pi)
print(math.cos(math.pi/7)) #we always have to refer the library any function/value comes from even if it is already nested inside a function of the same package
```

A way to shorten the calls to functions and variables stored in libraries is **importing specific items** alone. In that case the syntax is the following and there's no need to specify the library the functions/variables belong to.

```python
from math import cos, pi
print(pi)
print(cos(pi/7))
```

Another way to shorten the syntax is **renaming the library** with a shorten acronym so that the calls are easier:

```python
import math as m
print(m.pi)
print(m.cos(m.pi/7))
```
