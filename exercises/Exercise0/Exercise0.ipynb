**numpy arrays can be sliced (e.g taking a subset of elements)**

```python
a = np.arange(10)
a = a**2

print(f"a = {str(a)}")
print(f"the second element is a[2] : {a[2]}")
print(f"the last element is a[-1] : {a[-1]}")
print(f"the subset from the 2 to 5-1 element is a[2:5] : {a[2:5]}")
```

**Example with 2D array**

**Sometimes we need to add an additional axes to the numpy array, to make it "higher dimensional". This is usually needed when combining a tensor with other tensors for various operations**

```python
a = np.ones(10)
print(a)
print(np.expand_dims(a, axis=1))  # Tis is effectively creating a matrix with 1 column
print(np.expand_dims(a, axis=0))  # This is creating a matrix of 1 row only
```

This is equivalent to the syntax `[:,None]` which is used very often.

```python
a = np.ones(10)
print(a)
print(a[:, None])
print(a[None, :])
```

**Iterating in python**

```python
for row in b:
    print(row)  # nb, the tabular is important

# flattening the array first (i.e. going from 2D to 1D)
for element in b.flat:
    print(element)

# several ways of reshaping the array :
k = b.ravel()
print(f"raveled array: {k}")

# full reshaping:
b = b.reshape(9, 1)  # 9 elements on the first dimention, 1 on the second one
print(f"reshaped array\n{b}")
```

**How to stack arrays**

```python
a = np.array([(1, 2), (3, 4)])
b = np.array([(5, 6), (7, 8)])

# vertical stack
k = np.vstack((a, b))
print(k)

print("=====")

# horizontal stack
k = np.hstack((a, b))
print(k)
```

**Assignments and copies : <br>**
- assigning one array to the other does not make a full copy of the memory, as in c++
- For making a full copy, one has to make a 'deep copy'

```python
a = np.arange(12)
print(f"a = {a}")
b = a
print(f"b = {b}")
print(f"python remembers that b IS a: {b is a}")

# changing an element of b:
b[1] = -1
# check what happens on a
print(f"a after modification of b: {b}")

# now making a deep copy :
b = a.copy()
# changing an element of b
b[1] = 1234
print(f"a after modification of b: {a}, -> a has not been modified again")
```

### Plotting 2D grids

Often one needs to plot a grid of values corresponding to different x and y values. 
Here's a useful combination of matplotlib commands to plot this "2D grids"

```python
X = np.linspace(0, 1, 20)
Y = np.linspace(10,20, 20)
Z =  X[:,None] @ Y[None,:] # Just creating a 2D value map

# Converting the X, Y coordinates in a grid of values 
xgrid, ygrid = np.meshgrid(X,Y)
# We need to use the transpose of the values because the first dimension of Z varies along the rows, but 
# when we plot we usually interprete the first dimension as the x axis, which are the columns in terms of indices. 
plt.pcolor(xgrid, ygrid, np.transpose(Z))
plt.colorbar()
```

**Histograms with mathplotlib**

```python
# Fixing random state for reproducibility
np.random.seed(19680801)

mu, sigma = 100, 15
x = mu + sigma * np.random.randn(10000)

# the histogram of the data
n, bins, patches = plt.hist(
    x, 50, density=True, facecolor="g", alpha=0.75, ec="black"
)  # ec: edge color, the border of each bin

plt.xlabel("blah")
plt.ylabel("Probability")
plt.title("Histogram of blah")
plt.text(60, 0.025, r"$\mu=100,\ \sigma=15$")
plt.axis([40, 160, 0, 0.03])
plt.grid(True)
plt.show()
```

**Series indexing**

```python
d = {"a": 0.0, "b": 1.0, "c": 2.0, "d": 3.0}
sd = pd.Series(d)

print("the series takes the key of the dictionnary as index")
print(sd)
print(f"\nelement of sd given by sd[2]: {sd[2]}")
print(f"also accessible by the key sd['b']: {sd['b']}")
print(f"N.B. -1 counts from the end sd[-1]: {sd[-1]}")
print(f"and :-1 counts everything minus the last element sd[:-1] :\n{sd[:-1]}")
```
