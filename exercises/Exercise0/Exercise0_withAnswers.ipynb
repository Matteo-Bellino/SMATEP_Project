## Broadcasting

Documentation: https://numpy.org/doc/stable/user/basics.broadcasting.html

The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations. Subject to certain constraints, the smaller array is “broadcast” across the larger array so that they have compatible shapes. Broadcasting provides a means of vectorizing array operations so that looping occurs in C instead of Python. It does this without making needless copies of data and usually leads to efficient algorithm implementations. 

```python
A = np.ones((4,5))*2
B = np.ones(3)*3
print(f"A shape {A.shape}")
print(f"B shape {B.shape}")
```

```python
A * B
```

When operating on two arrays, NumPy compares their shapes element-wise. It starts with the trailing (i.e. rightmost) dimension and works its way left. Two dimensions are compatible when

-   they are equal, or
-   one of them is 1.

If these conditions are not met, a ValueError: operands could not be broadcast together exception is thrown, indicating that the arrays have incompatible shapes.

Input arrays do not need to have the same number of dimensions. The resulting array will have the same number of dimensions as the input array with the greatest number of dimensions, where the size of each dimension is the largest size of the corresponding dimension among the input arrays. Note that missing dimensions are assumed to have size one.

```python
A = np.ones((4,5,1))*2
B = np.ones(3)*3
print(f"A shape {A.shape}")
print(f"B shape {B.shape}")
(A*B).shape
```

```python
A = np.ones((4,5,1,6))*1
B = np.ones((3, 6))*3
print(f"A shape {A.shape}")
print(f"B shape {B.shape}")
(A*B).shape
```

### Plotting 2D grids

Often one needs to plot a grid of values corresponding to different x and y values. 
Here's a useful combination of matplotlib commands to plot this "2D grids"

```python
X = np.linspace(0, 1, 20)
Y = np.linspace(10,20, 20)
Z =  X[:,None] @ Y[None,:] # Just creating a 2D value map

# Converting the X, Y coordinates in a grid of values 
xgrid, ygrid = np.meshgrid(X,Y)
# We need to use the transpose of the values because the first dimension of Z varies along the rows, but 
# when we plot we usually interpret the first dimension as the x axis, which are the columns in terms of indices. 
plt.pcolor(xgrid, ygrid, np.transpose(Z))
plt.colorbar()
```

## Scenario 1

$$ p(mu|signal)  = \frac{p(sig|\mu)p(\mu)}{p(sig|\mu)p(\mu) + p(sig|\pi)p(\pi)}$$

By inverting the formula to extract  $p(signal|mu)$ the tagging efficiency we get:

$$p(signal|mu) = \frac{p(\mu|sig)p(sig|\pi)p(\pi)}{(1-p(\mu|sig))p(\mu)}$$


```python
p_mistag = np.linspace(0, 0.99, 50)
p_pi = 0.06
p_mu = 0.94
p_sig_pi = p_mistag
p_mu_sig = 0.95
p_sig_mu = p_mu_sig * p_sig_pi * p_pi / (( 1- p_mu_sig)*p_mu)
```

```python
plt.plot(p_mistag, p_sig_mu)
plt.xlabel("p_mistag")
plt.ylabel("p_tag = p(signal|mu)")
plt.text(0.5,0.5, "p(mu|signal) == 95%")
plt.plot([0,1], [1.,1.])
```

**Question: Do you observe any limitation of your detector if you want to reach this level efficiency?**: 

The limitation consists in a maximum value for the probability of tagging background as signal (mistag). If the mistag probability is larger than 0.808, it is impossible to reach the desidered level of probability of obserbing a muon given a signal (with the given beam components proportions).  This limit can be understood by checking the plot above. The $p_{tag}$ should be larger than 1 to reach the desidered efficiency and that's impossible.  

```python
print(f"max mistag: {np.max(p_mistag[p_sig_mu <=1]):.5f}")
```

## Scenario 2

Let's consider each layer of the detector has a new test that is updating our prior. 

After the first layer: 
$$ p(\mu|sig)  = \frac{p(sig|\mu)\,p(\mu)}{p(sig|\mu)\,p(\mu) + p(sig|\pi)\,p(\pi)}$$

Now $p(\mu|sig)$ becomes the prior probability for the measurement of the second layer
$$ p(\mu|sig_2, sig_1)  = \frac{p(sig_2|\mu)\,p(\mu|sig_1)}{p(sig_2|\mu)\,p(\mu) + p(sig_2|\pi)\,p(\pi)}$$

```python
def get_posterior(prior, p_tag, p_mistag):
    return p_tag * prior / ((p_tag*prior)+(p_mistag*(1-prior)))
```

```python
def get_n_layers(p_muon, p_tag, p_mistag):
    if p_tag < p_mistag: 
        print("Experiment is malformed p_tag < p_mistag")
        return 0
    prior = p_muon
    N_layers = 0
    
    while True: 
        N_layers += 1
        posterior = get_posterior(prior, p_tag, p_mistag)
        #print(prior , posterior)
        if posterior > 0.99:   
            break
        else:
            prior = posterior
        if N_layers > 50: break
    return N_layers
```

```python
p_mistag = 0.10
p_tag = 0.60
p_muon = 0.60
print(get_n_layers(p_muon, p_tag, p_mistag))
```

```python
p_tag = np.linspace(0.5,1, 20)
p_mistag = np.linspace(0.5,1, 20)
nlayers = np.zeros((20,20))

for t in range(20):
    for m in range(20):
        nlayers[t,m] = get_n_layers(p_muon, p_tag[t],p_mistag[m])
```

```python
X, y = np.meshgrid(p_tag, p_mistag)
plt.pcolor(X, y, nlayers.T)
plt.xlabel("p_tag")
plt.ylabel("p_mistag")
plt.colorbar()
```

Alternatively, *since measurements in different layers are independent*, one can write the probability of a particle, $\mu$ or $\pi$, to leave a signal in N consecutive layers as 
$$p(sig^{N}|\mu) = [p(sig|\mu)]^N\qquad\qquad p(sig^{N}|\pi) = [p(sig|\pi)]^N$$
and ultimately write a single Bayes equation such as
$$p(\mu|sig^N) = \frac{p(sig^N|\mu)\,p(\mu)}{p(sig^N)} = \frac{p(sig^N|\mu)\,p(\mu)}{p(sig^N|\mu)\,p(\mu) + p(sig^N|\pi)\,p(\pi)}  $$

```python
def prob_layers(n, prob_sig_given_mu, prob_sig_given_pi, frac_mu, frac_pi):
    return np.power(prob_sig_given_mu, n) * frac_mu / (np.power(prob_sig_given_mu, n) * frac_mu + np.power(prob_sig_given_pi, n) * frac_pi)

# print P(mu|sig^n) as a function of n
for n in range(5):
    print('signal present in %d consecutive layers - %.3f likely to be a muon' %(n, prob_layers(n, 0.6, 0.1, 0.6, 0.4)))
```
