# Broadcasting

Documentation: https://numpy.org/doc/stable/user/basics.broadcasting.html

The term broadcasting describes how NumPy treats arrays with different shapes during arithmetic operations. Subject to certain constraints, the smaller array is “broadcast” across the larger array so that they have compatible shapes. Broadcasting provides a means of vectorizing array operations so that looping occurs in C instead of Python. It does this without making needless copies of data and usually leads to efficient algorithm implementations. 

```python
A = np.ones((4,5))*2
B = np.ones(3)*3
print(f"A shape {A.shape}")
print(f"B shape {B.shape}")
```

```python
A * B
```

When operating on two arrays, NumPy compares their shapes element-wise. It starts with the trailing (i.e. rightmost) dimension and works its way left. Two dimensions are compatible when

-   they are equal, or
-   one of them is 1.

If these conditions are not met, a ValueError: operands could not be broadcast together exception is thrown, indicating that the arrays have incompatible shapes.

Input arrays do not need to have the same number of dimensions. The resulting array will have the same number of dimensions as the input array with the greatest number of dimensions, where the size of each dimension is the largest size of the corresponding dimension among the input arrays. Note that missing dimensions are assumed to have size one.

```python
A = np.ones((4,5,1))*2
#B = np.ones(3)*3
B = np.array((0.5, 1, 2))
print(f"A shape {A.shape}")
print(f"B shape {B.shape}")
(A*B).shape
#print(A*B)
```

```python
A = np.ones((4,5,1,6))*1
B = np.ones((3, 6))*3
print(f"A shape {A.shape}")
print(f"B shape {B.shape}")
(A*B).shape
```

```python
A = np.ones((4,5,1,6))*1
B = np.ones(( 5, 3, 6))*3
print(f"A shape {A.shape}")
print(f"B shape {B.shape}")
(A*B).shape
```

```python
A = np.ones((4,5,1,6))*1
B = np.ones(( 1, 5, 3, 6))*3
print(f"A shape {A.shape}")
print(f"B shape {B.shape}")
(A*B).shape
```

### Plotting 2D grids

Often one needs to plot a grid of values corresponding to different x and y values. 
Here's a useful combination of matplotlib commands to plot this "2D grids"

```python
X = np.linspace(0, 1, 20)
Y = np.linspace(10,20, 20)
Z =  X[:,None] @ Y[None,:] # Just creating a 2D value map, in this case it is simply the product of x and y

# Converting the X, Y coordinates in a grid of values 
xgrid, ygrid = np.meshgrid(X,Y) #it creates two grids in which x and y are repeated to match the number of entries of the other array
#print(X)
#print("####")
#print(xgrid)
# We need to use the transpose of the values because the first dimension of Z varies along the rows, but 
# when we plot we usually interpret the first dimension as the x axis, which are the columns in terms of indices. 
plt.pcolor(xgrid, ygrid, np.transpose(Z))
plt.colorbar()
```
